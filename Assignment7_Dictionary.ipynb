```python
#1 question
students=eval(input("Enter the dictionary of students marks : "))
for names,marks in students.items():
  print(names, ":" ,marks)
```

```python
#2 question
sentence = input("Enter a sentence: ")

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
#3 question
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print("Merged Dictionary:", result)
```
