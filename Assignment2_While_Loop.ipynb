```python
#1 question
n = int(input("Enter a number: "))
s = 0
while n > 0:
    digit = n % 10
    s = s + digit
    n = n // 10
print("Sum of digits =", s)
```

```python
#2 question
n = int(input("Enter a number: "))
r = 0
while n > 0:
    digit = n % 10
    r = r * 10 + digit
    n = n // 10
print("Reversed number =", r)
```

```python
#3 question
s = 0
while True:
    n = int(input("Enter a number: "))
    if n == 0:
        break
    s += n
print("Sum =", s)
```
